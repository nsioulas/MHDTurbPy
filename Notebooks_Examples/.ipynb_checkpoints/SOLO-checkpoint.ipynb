```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import welch, stft, istft, medfilt


# ============================================================
# Synthetic data: broadband + drifting wheel lines
# ============================================================
def make_powerlaw_noise(fs, n, alpha=2.7, seed=0):
    rng = np.random.default_rng(seed)
    f = np.fft.rfftfreq(n, d=1.0 / fs)
    amp = np.ones_like(f)
    amp[1:] = f[1:] ** (-alpha / 2.0)
    phase = rng.uniform(0, 2 * np.pi, size=f.size)
    X = amp * (np.cos(phase) + 1j * np.sin(phase))
    x = np.fft.irfft(X, n=n)
    x = (x - np.mean(x)) / np.std(x)
    return x


def build_synthetic_case(fs=256.0, duration_s=900.0, seed=2):
    rng = np.random.default_rng(seed)
    n = int(duration_s * fs)
    t = np.arange(n) / fs

    bg = make_powerlaw_noise(fs, n, alpha=2.7, seed=seed)

    # drifting + intermittent tones
    base_freqs = np.array([23.0, 47.0, 71.0, 94.0, 117.0])
    amps = np.array([0.35, 0.25, 0.20, 0.18, 0.15])

    wheel = np.zeros_like(t)
    for f0, a0 in zip(base_freqs, amps):
        # drift amplitude ~0.4 Hz over minutes (exaggerated but realistic-ish)
        drift = 0.01 * np.pi * (1 / 200.0) * t + 2 * np.pi * rng.random()
        f_inst = f0 + drift

        # intermittent envelope
        bursts = (rng.random(n) > 0.996).astype(float)
        bursts = np.convolve(bursts, np.ones(int(fs * 1.5)), mode="same")
        bursts = np.clip(bursts, 0, 1)
        env = 0.15 + 0.85 * bursts

        phase = 2 * np.pi * np.cumsum(f_inst) / fs
        wheel += a0 * env * np.cos(phase)

    x = bg + wheel + 0.05 * rng.normal(size=n)
    return t, x


# ============================================================
# Core method: STFT mask + attenuation (drift-safe)
# ============================================================
def interp_nans(x):
    x = np.asarray(x, float)
    n = x.size
    if n == 0:
        return x
    ok = np.isfinite(x)
    if ok.all():
        return x
    if ok.sum() < 2:
        return np.zeros_like(x)
    idx = np.arange(n)
    y = x.copy()
    y[~ok] = np.interp(idx[~ok], idx[ok], x[ok])
    return y


def dilate_mask(mask, df, dt, freq_pad_hz=0.3, time_pad_s=0.0):
    """
    Simple rectangular dilation without scipy.ndimage.
    """
    mf = mask.copy()

    # frequency dilation
    if freq_pad_hz > 0:
        pad_f = int(np.ceil(freq_pad_hz / df))
        if pad_f > 0:
            for _ in range(pad_f):
                mf[:-1, :] |= mf[1:, :]
                mf[1:, :] |= mf[:-1, :]

    # time dilation
    if time_pad_s > 0:
        pad_t = int(np.ceil(time_pad_s / dt))
        if pad_t > 0:
            for _ in range(pad_t):
                mf[:, :-1] |= mf[:, 1:]
                mf[:, 1:] |= mf[:, :-1]

    return mf


def remove_wheel_noise_dynamic_mask(
    x,
    fs,
    *,
    freq_min=8.0,
    stft_nperseg=4096,
    stft_overlap=0.5,
    kernel=151,
    percentile_q=97.0,
    thresh_db=6.0,
    mad_mult=5.0,
    freq_pad_hz=0.35,
    time_pad_s=0.0,
    atten_db=80.0,
    return_debug=False,
):
    """
    Drift-safe removal:
      - STFT -> S=|Z|^2
      - build score_db(f,t) relative to per-time baseline
      - mask bins above threshold (freq >= freq_min)
      - dilate mask in frequency/time
      - attenuate masked STFT coefficients
      - iSTFT
    """
    x = interp_nans(np.asarray(x, float))
    n = x.size

    nperseg = int(min(max(256, stft_nperseg), n))
    noverlap = int(nperseg * float(stft_overlap))
    noverlap = min(max(0, noverlap), nperseg - 1)

    f, tt, Z = stft(
        x, fs=fs, window="hann",
        nperseg=nperseg, noverlap=noverlap,
        boundary=None, padded=False
    )

    if f.size < 2 or Z.size == 0:
        return (x, None) if return_debug else x

    S = np.abs(Z) ** 2
    eps = 1e-30
    logS = np.log10(np.maximum(S, eps))  # (nf, nt)

    # Build baseline PER TIME slice: smooth across frequency
    k = int(kernel)
    if k < 3:
        k = 3
    if k % 2 == 0:
        k += 1
    if k > logS.shape[0]:
        k = logS.shape[0] if (logS.shape[0] % 2 == 1) else max(3, logS.shape[0] - 1)

    baseline = np.empty_like(logS)
    for j in range(logS.shape[1]):
        baseline[:, j] = medfilt(logS[:, j], kernel_size=k)

    score_db = 10.0 * (logS - baseline)

    # Robust threshold per time-slice (handles variable background)
    mask = np.zeros_like(score_db, dtype=bool)
    freq_mask = f >= float(freq_min)

    for j in range(score_db.shape[1]):
        s = score_db[:, j]
        s = s[freq_mask]

        if s.size == 0:
            continue

        med = float(np.median(s))
        mad = float(np.median(np.abs(s - med)))
        sigma = 1.4826 * mad if mad > 0 else 0.0

        thr = float(thresh_db)
        if sigma > 0:
            thr = max(thr, med + float(mad_mult) * sigma)

        m = (score_db[:, j] > thr) & freq_mask
        mask[:, j] = m

    # Dilate to catch drift/leakage
    df = float(f[1] - f[0])
    dt = float(tt[1] - tt[0]) if tt.size > 1 else (nperseg / fs)
    mask2 = dilate_mask(mask, df=df, dt=dt, freq_pad_hz=freq_pad_hz, time_pad_s=time_pad_s)

    # Attenuate masked coefficients
    atten = 10.0 ** (-float(atten_db) / 20.0)
    Z_clean = Z.copy()
    Z_clean[mask2] *= atten

    _, y = istft(
        Z_clean, fs=fs, window="hann",
        nperseg=nperseg, noverlap=noverlap,
        boundary=None
    )

    if y.size > n:
        y = y[:n]
    elif y.size < n:
        y = np.pad(y, (0, n - y.size))

    if return_debug:
        debug = {"f": f, "t": tt, "mask": mask2, "score_db": score_db}
        return y, debug

    return y


# ============================================================
# Run MRE and plot
# ============================================================
if __name__ == "__main__":
    fs = 256.0
    t, x = build_synthetic_case(fs=fs, duration_s=9000.0, seed=3)

    x_clean, dbg = remove_wheel_noise_dynamic_mask(
        x, fs,
        freq_min=8.0,
        stft_nperseg=4096/2,
        stft_overlap=0.5,
        kernel=151,
        percentile_q=97.0,  # (kept for consistency; not used in this time-slice method)
        thresh_db=2.0,
        mad_mult=5.0,
        freq_pad_hz=0.85,   # IMPORTANT for drift/leakage
        time_pad_s=0.5,
        atten_db=80.0,
        return_debug=True
    )

    f_raw, P_raw = welch(x, fs=fs, window="hann", nperseg=2**15, noverlap=2**14, detrend="constant", scaling="density")
    f_cln, P_cln = welch(x_clean, fs=fs, window="hann", nperseg=2**15, noverlap=2**14, detrend="constant", scaling="density")

    f = dbg["f"]
    tt = dbg["t"]
    mask = dbg["mask"]
    score_db = dbg["score_db"]

    fig = plt.figure(figsize=(12, 8))

    ax1 = plt.subplot(2, 2, 1)
    ax1.loglog(f_raw, P_raw, lw=1.5, label="Raw")
    ax1.loglog(f_cln, P_cln, lw=1.5, label="Cleaned")
    ax1.set_xlabel(r"$f\,[\mathrm{Hz}]$")
    ax1.set_ylabel("PSD")
    ax1.set_title("Welch PSD (before/after)")
    ax1.set_xlim(1e-2, fs/2)
    ax1.legend(loc="best")

    ax2 = plt.subplot(2, 2, 2)
    ax2.pcolormesh(tt, f, score_db, shading="auto")
    ax2.set_ylim(0, 140)
    ax2.set_xlabel(r"$t\,[\mathrm{s}]$")
    ax2.set_ylabel(r"$f\,[\mathrm{Hz}]$")
    ax2.set_title("score_db(f,t)")

    ax3 = plt.subplot(2, 2, 3)
    ax3.pcolormesh(tt, f, mask.astype(float), shading="auto")
    ax3.set_ylim(0, 140)
    ax3.set_xlabel(r"$t\,[\mathrm{s}]$")
    ax3.set_ylabel(r"$f\,[\mathrm{Hz}]$")
    ax3.set_title("Detected mask (after dilation)")

    ax4 = plt.subplot(2, 2, 4)
    # Show raw STFT power for reference
    f0, tt0, Z0 = stft(x, fs=fs, window="hann", nperseg=2048, noverlap=1024)
    S0 = 10*np.log10(np.maximum(np.abs(Z0)**2, 1e-30))
    ax4.pcolormesh(tt0, f0, S0, shading="auto")
    ax4.set_ylim(0, 140)
    ax4.set_xlabel(r"$t\,[\mathrm{s}]$")
    ax4.set_ylabel(r"$f\,[\mathrm{Hz}]$")
    ax4.set_title("Raw STFT spectrogram (for intuition)")

    plt.tight_layout()
    plt.show()
```
